```python
import pandas as pd

df = pd.read_csv('../cleaned_data/youtube_tech_channels_cleaned.csv')
df.head()
```

```python
correlation = df['Duration Minutes'].corr(df['View Count'])
print(f"Correlation between Duration and Views: {correlation:.3f}")
```

```python
correlation2 = df['Tags Count'].corr(df['View Count'])
print(f"Correlation between Tags Count and Views: {correlation2:.3f}")
```

```python
df['Title Length'] = df['Title'].str.len()
correlation3 = df['Title Length'].corr(df['View Count'])
print(f"Correlation between Title Length and Views: {correlation3:.3f}")

df.groupby(pd.cut(df['Title Length'], bins=[0, 40, 60, 100]))['View Count'].mean()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(df['Duration Minutes'], df['View Count'], alpha=0.5, color='steelblue')
plt.xlabel('Duration (minutes)')
plt.ylabel('View Count')
plt.title('Video Duration vs Views')
plt.show()
```
